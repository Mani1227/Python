```python
#Create a barchart for HR Zone vs count and apply hatches to the bars.
zone_counts = df_test_measure['HR'].value_counts()
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(zone_counts.index, zone_counts.values, color='skyblue', edgecolor='black')
hatches = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']  # Different patterns
for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)
ax.set_xlabel('HR')
ax.set_ylabel('Count')
ax.set_title('HR vs Count with Hatches')
plt.tight_layout()
plt.show()
```
