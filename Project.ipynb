```python
import psycopg2
from psycopg2 import OperationalError

try:
    conn = psycopg2.connect(
        host="localhost",
    database="postgres",
    user="postgres",
    password="1234",
    port="5432"
    )
    # Check connection status
    if conn.status == psycopg2.extensions.STATUS_READY:
        print("Connection established successfully!")
    else:
        print("Connection status:", conn.status);
except OperationalError as e:
    print("The connection failed.")
    print(e)
```

```python
x= conn.cursor()
```

```python
x.execute("""
CREATE TABLE IF NOT EXISTS df_subject_info (
    participant_id INT PRIMARY KEY,
    name VARCHAR(100),
    age INT
);
""")
```

```python
x.execute("""
CREATE TABLE IF NOT EXISTS test_measure (
    participant_id INT,
    temperature FLOAT,
    humidity FLOAT
);
""")
```

```python
x.execute("""
SELECT table_name FROM information_schema.tables
WHERE table_schema='public';
""")

tables = x.fetchall()
print("Tables:", tables)
```

```python
# Insert data
for index, row in df_subject_info.iterrows():
    x.execute("""
    INSERT INTO df_subject_info (participant_id, name, age)
    VALUES (%s, %s, %s)
    ON CONFLICT (participant_id) DO NOTHING;
    """, (row['ID'], row['ID_test'], row['Age']))

for index, row in df_test_measure.iterrows():
    x.execute("""
    INSERT INTO df_test_measure (participant_id, temperature, humidity)
    VALUES (%s, %s, %s);
    """, (row['ID'], row['temperature'], row['humidity']))

conn.commit()
```

```python
cur = conn.cursor()

# Drop tables if they exist
cur.execute("DROP TABLE IF EXISTS subject_info CASCADE;")
cur.execute("DROP TABLE IF EXISTS test_measure CASCADE;")

conn.commit()
cur.close()
conn.close()

print("Tables dropped successfully.")
```
